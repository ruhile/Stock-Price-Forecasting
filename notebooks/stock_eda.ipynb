# 📈 Week 16 – Day 2: Exploratory Data Analysis (EDA)
## Apple Inc. (`AAPL`) Stock Price Analysis

This notebook performs in-depth Exploratory Data Analysis (EDA) on Apple Inc. (`AAPL`) stock data, analyzing price trends, trading volume, high-low spreads, daily return distributions, moving averages (MA20 & MA50), rolling volatility, and feature correlation heatmaps.

### 📚 Step 1: Load Libraries

```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

### 📥 Step 2: Load the Dataset

```python
stock = yf.download("AAPL", start="2018-01-01", end="2026-01-01")
if isinstance(stock.columns, pd.MultiIndex):
    stock = stock.xs("AAPL", axis=1, level=1) if "AAPL" in stock.columns.levels[1] else stock

stock.head()
```

### 📊 Step 3: Closing Price Trend

```python
plt.figure(figsize=(14, 6))
plt.plot(stock["Close"], color="blue")
plt.title("Apple Closing Price")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.show()
```

### 📈 Step 4: Trading Volume

```python
plt.figure(figsize=(14, 5))
plt.bar(stock.index, stock["Volume"])
plt.title("Daily Trading Volume")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.show()
```

### 📉 Step 5: High vs Low Prices

```python
plt.figure(figsize=(14, 6))
plt.plot(stock["High"], label="High")
plt.plot(stock["Low"], label="Low")
plt.title("Daily High and Low Prices")
plt.legend()
plt.show()
```

### 📌 Step 6: Daily Returns

```python
stock["Daily Return"] = stock["Close"].pct_change(fill_method=None)
stock.head()
```

### 📊 Step 7: Plot Daily Returns

```python
plt.figure(figsize=(14, 6))
plt.plot(stock["Daily Return"])
plt.title("Daily Returns")
plt.show()
```

### 📉 Step 8: Distribution of Daily Returns

```python
plt.figure(figsize=(10, 5))
sns.histplot(stock["Daily Return"].dropna(), bins=50, kde=True)
plt.title("Distribution of Daily Returns")
plt.show()
```

### 📈 Step 9: 20-Day Moving Average

```python
stock["MA20"] = stock["Close"].rolling(window=20).mean()

plt.figure(figsize=(14, 6))
plt.plot(stock["Close"], label="Close")
plt.plot(stock["MA20"], label="20-Day MA")
plt.legend()
plt.title("Closing Price vs 20-Day Moving Average")
plt.show()
```

### 📈 Step 10: 50-Day Moving Average

```python
stock["MA50"] = stock["Close"].rolling(50).mean()

plt.figure(figsize=(14, 6))
plt.plot(stock["Close"], label="Close")
plt.plot(stock["MA20"], label="MA20")
plt.plot(stock["MA50"], label="MA50")
plt.legend()
plt.title("Moving Averages")
plt.show()
```

### 📊 Step 11: Rolling Standard Deviation (Volatility)

```python
stock["Rolling Std"] = stock["Close"].rolling(20).std()

plt.figure(figsize=(14, 6))
plt.plot(stock["Rolling Std"])
plt.title("20-Day Rolling Standard Deviation")
plt.show()
```

### 📋 Step 12: Correlation Matrix

```python
plt.figure(figsize=(8, 6))
sns.heatmap(stock.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Feature Correlation")
plt.show()
```
