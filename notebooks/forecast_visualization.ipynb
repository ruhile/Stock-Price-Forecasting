# 📈 Week 16 – Day 6: Visualization & Forecast Analysis
## Comprehensive Model Evaluation & Diagnostic Plots

This notebook covers:
- Loading prediction results & actual stock values
- Plotting Actual vs. Predicted time series trends
- Generating Scatter plots for goodness of fit
- Residual error analysis ($Residual = Actual - Predicted$)
- Residual distribution histogram & KDE curve
- 30-Day future forecast visualization
- Exporting summary performance metrics to `report/model_performance.csv`.

### 📥 Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from statsmodels.tsa.arima.model import ARIMA

plt.style.use("ggplot")
```

### 📂 Step 2: Load Prediction Results

```python
results = pd.read_csv("../data/predictions.csv", index_col=0, parse_dates=True)
results.head()
```

### 📈 Step 3: Actual vs Predicted Line Plot

```python
plt.figure(figsize=(14, 6))
plt.plot(results["Actual"], label="Actual")
plt.plot(results["Predicted"], label="Predicted")
plt.title("Actual vs Predicted Closing Price")
plt.xlabel("Time")
plt.ylabel("Price")
plt.legend()
plt.savefig("../images/actual_vs_predicted.png", dpi=300, bbox_inches="tight")
plt.show()
```

### 📉 Step 4: Scatter Plot

```python
plt.figure(figsize=(7, 7))
plt.scatter(results["Actual"], results["Predicted"], alpha=0.6)
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted Scatter")
plt.savefig("../images/actual_vs_predicted_scatter.png", dpi=300, bbox_inches="tight")
plt.show()
```

### 📌 Step 5 & 6: Residuals Calculation & Plot

```python
results["Residual"] = results["Actual"] - results["Predicted"]

plt.figure(figsize=(14, 5))
plt.plot(results["Residual"])
plt.axhline(y=0, color="red", linestyle="--")
plt.title("Residual Plot")
plt.savefig("../images/residual_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```

### 📈 Step 7: Distribution of Residuals

```python
plt.figure(figsize=(10, 5))
sns.histplot(results["Residual"], bins=40, kde=True)
plt.title("Residual Distribution")
plt.savefig("../images/residual_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

### 📉 Step 8: Future Forecast Plot

```python
stock = pd.read_csv("../data/processed_stock_data.csv", index_col="Date", parse_dates=True)
close_prices = stock["Close"]
final_model = ARIMA(close_prices, order=(5, 1, 0))
final_fit = final_model.fit()
future_forecast = final_fit.forecast(steps=30)

plt.figure(figsize=(14, 6))
plt.plot(close_prices, label="Historical")
plt.plot(future_forecast.index, future_forecast, label="Forecast")
plt.legend()
plt.title("30-Day Future Forecast")
plt.savefig("../images/future_forecast_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```

### 📊 Step 9 & 10: Performance Summary

```python
mae = mean_absolute_error(results["Actual"], results["Predicted"])
rmse = np.sqrt(mean_squared_error(results["Actual"], results["Predicted"]))
r2 = r2_score(results["Actual"], results["Predicted"])

summary = pd.DataFrame({
    "Metric": ["MAE", "RMSE", "R² Score"],
    "Value": [mae, rmse, r2]
})

summary.to_csv("../report/model_performance.csv", index=False)
summary
```
