# 📈 Week 16 – Day 5: Time Series Forecasting with ARIMA
## AutoRegressive Integrated Moving Average (ARIMA)

This notebook covers end-to-end time series modeling with ARIMA(5,1,0):
- Understanding ARIMA components ($p$, $d$, $q$)
- 80/20 train-test split
- Fitting ARIMA model on historical training data
- Evaluating test predictions (MAE, RMSE)
- Visualizing full train/test/forecast curves
- Fitting on full dataset & forecasting next 30 trading days
- Model comparison: Linear Regression vs ARIMA.

### 📥 Step 1 & 2: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_error, mean_squared_error

plt.style.use("ggplot")
```

### 📊 Step 3: Load Dataset

```python
stock = pd.read_csv("../data/processed_stock_data.csv", index_col="Date", parse_dates=True)
close_prices = stock["Close"]
stock.head()
```

### ✂️ Step 4: Train-Test Split

```python
train_size = int(len(close_prices) * 0.8)
train = close_prices.iloc[:train_size]
test = close_prices.iloc[train_size:]

print("Train size:", len(train))
print("Test size:", len(test))
```

### 📈 Step 5: Plot Train & Test Data

```python
plt.figure(figsize=(14, 6))
plt.plot(train, label="Train")
plt.plot(test, label="Test")
plt.title("Train-Test Split")
plt.legend()
plt.savefig("../images/arima_train_test.png", dpi=300)
plt.show()
```

### 📌 Step 6 & 7: Build & Fit ARIMA Model, Forecast Test Period

```python
model = ARIMA(train, order=(5, 1, 0))
model_fit = model.fit()

forecast = model_fit.forecast(steps=len(test))
forecast.index = test.index
```

### 📊 Step 8: Evaluate the Model

```python
mae = mean_absolute_error(test, forecast)
rmse = np.sqrt(mean_squared_error(test, forecast))

print("MAE: ", mae)
print("RMSE:", rmse)
```

### 📈 Step 9: Plot Forecast

```python
plt.figure(figsize=(14, 6))
plt.plot(train, label="Train", color="blue")
plt.plot(test, label="Actual", color="orange")
plt.plot(test.index, forecast, label="Forecast", color="green", linestyle="--")
plt.legend()
plt.title("ARIMA Forecast")
plt.savefig("../images/arima_full_forecast.png", dpi=300)
plt.show()
```

### 📌 Step 10: Forecast the Next 30 Days

```python
final_model = ARIMA(close_prices, order=(5, 1, 0))
final_fit = final_model.fit()

future_forecast = final_fit.forecast(steps=30)
print(future_forecast)
```

### 📈 Step 11: Visualize Future Forecast

```python
plt.figure(figsize=(14, 6))
plt.plot(close_prices, label="Historical Data", color="blue")
plt.plot(future_forecast.index, future_forecast, label="30-Day Forecast", color="red", linestyle="--")
plt.legend()
plt.title("30-Day Stock Price Forecast")
plt.savefig("../images/arima_30day_forecast.png", dpi=300)
plt.show()
```

### 📊 Step 12: Compare Linear Regression vs ARIMA

| Model | Advantages | Limitations |
| :--- | :--- | :--- |
| **Linear Regression** | Simple, fast, highly interpretable with multi-feature support | Doesn't naturally model temporal sequence dependencies |
| **ARIMA** | Designed specifically for time series, captures autoregressive and trend patterns | Requires stationary data and parameter tuning; long multi-step forecasts converge toward mean |
