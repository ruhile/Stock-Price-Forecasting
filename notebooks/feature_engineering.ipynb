# 📈 Week 16 – Day 3: Feature Engineering for Time Series
## Apple Inc. (`AAPL`) Stock Price Feature Creation

This notebook builds a comprehensive set of temporal features for stock price forecasting, including:
- **Lag Features**: Lag_1, Lag_2, Lag_3, Lag_5, Lag_10
- **Price Moving Averages**: MA10, MA20, MA50
- **Price Volatility**: STD20
- **Return Metrics**: Daily_Return, Return_MA10, Return_STD10
- **Data Cleaning**: Missing value removal and export to `processed_stock_data.csv`.

### 📥 Step 1: Load Dataset

```python
import yfinance as yf
import pandas as pd
import numpy as np

stock = yf.download("AAPL", start="2018-01-01", end="2026-01-01")
if isinstance(stock.columns, pd.MultiIndex):
    stock = stock.xs("AAPL", axis=1, level=1) if "AAPL" in stock.columns.levels[1] else stock

stock.head()
```

### 📌 Step 2: Create Lag Features

```python
# 1-Day Lag
stock["Lag_1"] = stock["Close"].shift(1)
# 2-Day Lag
stock["Lag_2"] = stock["Close"].shift(2)
# 3-Day Lag
stock["Lag_3"] = stock["Close"].shift(3)
# 5-Day Lag
stock["Lag_5"] = stock["Close"].shift(5)
# 10-Day Lag
stock["Lag_10"] = stock["Close"].shift(10)
```

### 📊 Step 3: View the New Features

```python
stock.head(15)
```

### 📈 Step 4: Moving Averages

```python
# 10-Day Moving Average
stock["MA10"] = stock["Close"].rolling(10).mean()
# 20-Day Moving Average
stock["MA20"] = stock["Close"].rolling(20).mean()
# 50-Day Moving Average
stock["MA50"] = stock["Close"].rolling(50).mean()
```

### 📉 Step 5: Rolling Standard Deviation

```python
stock["STD20"] = stock["Close"].rolling(20).std()
```

### 📈 Step 6: Daily Percentage Return

```python
stock["Daily_Return"] = stock["Close"].pct_change(fill_method=None)
```

### 📊 Step 7: Rolling Average of Daily Returns

```python
stock["Return_MA10"] = stock["Daily_Return"].rolling(10).mean()
```

### 📉 Step 8: Rolling Volatility of Returns

```python
stock["Return_STD10"] = stock["Daily_Return"].rolling(10).std()
```

### 📋 Step 9: Check Missing Values

```python
print("Missing values created by rolling/lag calculations:")
print(stock.isnull().sum())
```

### 🧹 Step 10: Remove Missing Values

```python
stock.dropna(inplace=True)
print("Missing values after dropna:")
print(stock.isnull().sum())
```

### 🔍 Step 11: View Final Dataset

```python
display(stock.head())
print("Final Shape:", stock.shape)
print("Final Columns:", list(stock.columns))
```

### 💾 Step 12: Save the Processed Dataset

```python
stock.to_csv("../data/processed_stock_data.csv")
print("Saved dataset to ../data/processed_stock_data.csv")
```
